```python
import pandas as pd

ifs_raw = pd.read_csv("ifs_two_child_limit_savings.csv", encoding="utf-8-sig")
ifs = (ifs_raw.rename(columns={
            "Financial year": "year",
            "Number of households affected ('000s)": "households_thousand",
            "Number of children affected ('000s)": "children_thousand",
            "Fiscal saving (£ billion, 2024/25 prices)": "cost_bn"})
       .query("year != 'Full rollout'"))
ifs.to_csv("ifs_two_child_limit_cost.csv", index=False)

obr_raw = pd.read_csv("obr_two_child_limit_savings.csv", encoding="utf-8-sig")
row = obr_raw[obr_raw.iloc[:, 0] == "Post-behavioural costing"].iloc[0, 1:]
obr = pd.DataFrame({"year": [c.replace("-", "/") for c in row.index],
                    "post_behavioural_cost_bn": row.astype(float).values})
obr.to_csv("obr_two_child_limit_costing.csv", index=False)
```
