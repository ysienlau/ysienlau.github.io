```python
import pandas as pd, requests

tfr = pd.read_csv("tfr_worldbank.csv")
tfr = tfr.rename(columns={"Country Name": "country", "Country Code": "code",
                          "Year": "year", "Value": "value"})[["country", "code", "year", "value"]]

meta = requests.get("https://api.worldbank.org/v2/country",
                    params={"format": "json", "per_page": 400}).json()[1]
meta = pd.DataFrame([{"code": c["id"], "region": c["region"]["value"],
                      "income": c["incomeLevel"]["value"]} for c in meta])

df = tfr.merge(meta, on="code", how="left")
print(df[df["region"].isna()]["country"].unique())   # anything that failed to match
print(df["region"].unique())                         # the exact region names

df[df["region"] != "Aggregates"].to_csv("tfr_countries.csv", index=False)
df[df["region"] == "Aggregates"].to_csv("tfr_aggregates.csv", index=False)
```

```python
df_2024 = df[df['year'] == 2024]
df_2024 = df_2024[df_2024['region']!='Aggregates']
```

```python
df_1990 = df[df["year"] == 1990]
df_1990 = df_1990[df_1990["region"] != "Aggregates"].dropna(subset=["value"])

lowest_1990 = df_1990.nsmallest(30, "value")
highest_1990 = df_1990.nlargest(30, "value")

print(len(df_1990), len(lowest_1990), len(highest_1990))
```

```python
lowest_1990
```

```python
highest_1990
```

```python
frames = [
    lowest_1990.assign(group="Lowest 30", snapshot_year=1990),
    highest_1990.assign(group="Highest 30", snapshot_year=1990),
    lowest_2024.assign(group="Lowest 30", snapshot_year=2024),
    highest_2024.assign(group="Highest 30", snapshot_year=2024),
]
```

```python
extremes_all = (pd.concat(frames)[["country", "code", "region", "value", "group", "snapshot_year"]]
                .rename(columns={"value": "tfr"}))
```

```python
extremes_all["region"] = extremes_all["region"].replace(
    {"Middle East, North Africa, Afghanistan & Pakistan": "MENA, Afghanistan & Pakistan"})
```

```python
print(extremes_all.groupby(["snapshot_year", "group"]).size())
print(pd.crosstab([extremes_all["snapshot_year"], extremes_all["group"]], extremes_all["region"]))
```

```python
extremes_all.to_csv("tfr_extremes_years.csv", index=False)
```
